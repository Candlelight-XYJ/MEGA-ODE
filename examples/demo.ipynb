# MEGA-ODE demo\n\nThis notebook demonstrates the package API. The core implementation lives in `src/megaode/`, not in this notebook.\n

```python
from megaode import MEGAODE, load_demo_data, evaluate_prediction\n
```

```python
data = load_demo_data("../demo_data", max_nodes=300)\nlen(data), data.num_features, data[0].num_nodes\n
```

```python
model = MEGAODE(\n    in_feats=data.num_features,\n    hidden_size=16,\n    num_experts=2,\n    out_feats=data.num_features,\n    time_tick_num=5,\n    gate_hidden_dim=16,\n)\n
```

```python
model.fit(data, epochs=5, lr=1e-3, patience=None)\n
```

```python
pred = model.predict(data, input_key="x36", output_index=3)\ntruth = [graph.ndata["y96"] for graph in data.graphs]\nevaluate_prediction(pred[0].squeeze(-1), truth[0].squeeze(-1))\n
```
